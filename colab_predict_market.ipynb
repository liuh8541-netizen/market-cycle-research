```python
import sys
import subprocess
from pathlib import Path

forecast_date = input('請輸入日期，例如 2026-07-14：').strip()
if not forecast_date:
    raise ValueError('未輸入日期')

cmd = [
    sys.executable,
    'scripts/predict_market.py',
    '--date', forecast_date,
    '--output', 'reports/today_market_forecast_colab.md',
    '--json-output', 'reports/today_market_forecast_colab.json',
]

result = subprocess.run(cmd, text=True, capture_output=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise SystemExit(result.returncode)

report_path = Path('reports/today_market_forecast_colab.md')
print('\n報告檔案：', report_path.resolve())
```
